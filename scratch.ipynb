```python
import torch
from torch.nn import functional as F
```

```python
B,T,C = 4, 8, 2

x = torch.randn((B,T,C))

x_bag_of_words = torch.zeros((B,T,C))

for batch in range (B):
    for time in range(T):

        # We take the previous time steps and the current one, then average it
        xprev = x[batch, :time+1]

        #We average over the time dimension, so 
        x_bag_of_words[batch, time] = torch.mean(xprev, dim=0) #Meaning crushes (t,c) into (c)

print(x[0])
print(x_bag_of_words[0])



```
